```python
import yfinance as yf
from datetime import datetime
import matplotlib.pyplot as plt
import pandas as pd
```

```python
airtel = yf.Ticker("BHARTIARTL.BO")
data = airtel.history(period="5y")  # last 1 year
data
```

```python
data.index=pd.to_datetime(data.index)
```

```python
data.tail()
```

```python
plt.figure(figsize=(15,5))
data['Close'].plot()
plt.title("airtel share Trend")
```

```python
plt.figure(figsize=(15,5))
data['Close'].plot(color = 'green')
plt.xlim("2023-01-01","2026-04-01")
plt.title("airtel share Trend 2024-01-01  to 2026-03-01")
plt.show()
```

# Time Resampling

```python
plt.figure(figsize=(15,5))
data.resample('MS').max()['Close'].plot()
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(15,5))
data['Close'].plot()
data['Close'].rolling(14).mean().plot()
plt.title("airtel share Trend")
```

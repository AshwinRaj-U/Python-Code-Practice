# Time Resampling

```python
data.resample('MS').median()
```

```python
plt.figure(figsize=(15,5))
data.resample('MS').max()['Close'].plot()
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(15,5))
data['Close'].plot()
data['Close'].rolling(14).mean().plot()
plt.title("airtel share Trend")
```

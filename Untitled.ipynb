# Time Resampling

```python
data=data.resample('W').mean()
```

```python
plt.figure(figsize=(15,5))
data['Close'].plot()
data['Close'].rolling(50).mean().plot()
data['Close'].rolling(200).mean().plot()
plt.xlim("2024-10-01","2026-04-01")
plt.title("airtel share Trend")
plt.show()
```
